```python
import pandas as pd
import json

file_path = '/content/sample_data/schedule.xlsx'
df_raw = pd.read_excel(file_path, header=[0,1]) #첫번째 행(날짜), 두번째 행(요일) 불러옴


new_cols = []
for a, b in df_raw.columns:
    a_str = str(a).strip()
    # 날짜(숫자) 컬럼이면 숫자만 추출, 그 외(이름·요약)는 그대로 사용
    if a_str.isdigit():
        col = str(int(a_str))
    else:
        col = a_str
    new_cols.append(col)
df_raw.columns = new_cols


name_col = [c for c in df_raw.columns if c == '이름'][0] #'이름'인 항목의 열
date_cols = [c for c in df_raw.columns if c.isdigit()] #숫자로만 이루어진 날짜 열

# schedule 이중 딕셔너리 생성
schedule = {}
for _, row in df_raw.iterrows():
    name = row[name_col] #근무자 이름
    schedule[name] = {}
    for d in date_cols: #날짜 열 하나씩 읽어 근무타입 인식
        code = row[d]
        if pd.isna(code) or str(code).strip() in ['-', '–', '']:
            continue
        schedule[name][d] = str(code).strip()

#json 파일으로 추출
with open('schedule.json', 'w', encoding='utf-8') as f:
    json.dump(schedule, f, ensure_ascii=False, indent=2)
print("schedule.json 파일로 저장되었습니다.")
```
